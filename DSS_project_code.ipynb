### Loading the Dataset

In this step, we load the dataset into a pandas DataFrame using the `read_csv()` function.
The dataset is stored as a CSV file, which is a common format for tabular data.

We use the parameter `encoding="latin-1"` because some datasets contain special characters that cannot be read using the default UTF-8 encoding. This ensures the file is loaded without errors.

After loading the dataset, we display the first few rows using `df.head()` to understand the structure of the data, including column names and sample values.

This step is important because it allows us to verify that the dataset has been loaded correctly before starting data cleaning and analysis.


```python
df = pd.read_csv("data.csv", encoding="latin-1")

df.head()
```

### Understanding the Dataset Structure

In this step, we examine the structure of the dataset using the `df.info()` function.

This function provides important information such as:

* The total number of rows and columns
* The names of each column
* The number of non-null (non-missing) values
* The data type of each column (e.g., integer, float, object)

This helps us identify potential issues such as missing values or incorrect data types that need to be addressed during the data cleaning process.


```python
df.info()
```

### Checking for Missing Values

In this step, we check for missing values in each column using the `isnull().sum()` function.

Missing values are common in real-world datasets and must be handled properly during data cleaning. Identifying these values helps us decide how to treat them, such as removing rows or filling in missing data.

This step is essential to ensure data quality before performing analysis.


```python
df.isnull().sum()
```

### Handling Missing Values

In this step, we handle missing values in the dataset.

The `CustomerID` column contains a large number of missing values. Since customer identification is important for analysis, rows with missing `CustomerID` are removed.

This ensures that the dataset contains only valid customer transactions, improving the quality and reliability of the analysis.


```python
df = df.dropna(subset=["CustomerID"])

df.isnull().sum()
```

### Removing Cancelled Orders

In this step, we remove cancelled transactions from the dataset.

Cancelled orders are identified by an Invoice Number that starts with the letter "C". These transactions do not represent completed sales and may negatively affect analysis results.

Therefore, we filter out these records to ensure that only valid and completed transactions are included in the dataset.


```python
df = df[~df["InvoiceNo"].str.startswith("C")]

df.head()
```

### Handling Negative Quantities

In this step, we remove records with negative values in the `Quantity` column.

Negative quantities typically represent returned or refunded items, which do not reflect actual sales transactions. Including them may distort analysis results.

Therefore, we filter the dataset to keep only records with positive quantities, ensuring accurate sales analysis.


```python
df = df[df["Quantity"] > 0]

df.describe()
```

### Converting Invoice Date to Datetime

In this step, we convert the `InvoiceDate` column from a text format to a datetime format.

Datetime conversion is important because it allows us to perform time-based analysis such as identifying trends over time, grouping data by months or years, and creating time-series visualizations.

This transformation improves the dataset and prepares it for further analysis and visualization.


```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df.dtypes
```

### Removing Duplicate Records

In this step, we check for duplicate rows in the dataset and remove them.

Duplicate records can occur due to data entry errors or system issues and may lead to incorrect analysis results.

By removing duplicates, we ensure that each transaction is represented only once, improving the accuracy and reliability of the dataset.


```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

df.shape
```

### Removing Invalid Prices

In this step, we remove records where the `UnitPrice` is equal to zero.

A unit price of zero does not represent a valid sale and may indicate errors or incomplete transactions in the dataset.

Removing these records helps improve the accuracy of the analysis by ensuring that only valid sales data is included.


```python
df = df[df["UnitPrice"] > 0]

df.describe()
```

### Creating Total Price Feature

In this step, we create a new column called `TotalPrice`, which represents the total value of each transaction.

The total price is calculated by multiplying the quantity of items purchased by the unit price.

This new feature is important for analyzing revenue, identifying top-selling products, and generating business insights.


```python
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

df.head()
```

### Saving the Cleaned Dataset

In this step, we save the cleaned dataset to a new CSV file.

This file will be used for further analysis and visualization in tools such as Power BI.

Saving the cleaned data ensures that all preprocessing steps are preserved and allows us to work efficiently without repeating the cleaning process.


```python
df.to_csv("cleaned_ecommerce.csv", index=False)
```
